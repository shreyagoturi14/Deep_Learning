```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset=pd.read_csv(r"C:\Users\LENOVO\Downloads\placement (2).csv")
```

```python
dataset.head(3)
```

```python
# dataset=dataset.drop(columns=['Unnamed: 0'])
dataset=dataset.drop(dataset.columns[0],axis=1)
dataset.head()
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x="cgpa",y="iq",data=dataset,hue="placement")
plt.show()
```

```python
x=dataset.iloc[:,:-1]
y=dataset["placement"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=40)
```

```python
from sklearn.linear_model import Perceptron
```

```python
pr=Perceptron()
pr.fit(x_train,y_train)
```

```python
pr.score(x_train,y_train)*100,pr.score(x_test,y_test)*100
```

```python
dataset.head(2)
```

```python
plt.figure(figsize=(4,3))
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=pr)
plt.show()
```
